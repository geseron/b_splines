```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import os
import pickle
import ast
from pathlib import Path
```

```python
def make_dataframe(file_names: list, folder: str, sep: str = ';') -> pd.DataFrame:
    data_list = []
    for group_files in file_names:
        if group_files[0].endswith('.pkl'):
            for file in group_files:
                with open(folder + "/" + file, "rb") as f:
                    rows = pickle.load(f)
                data_list.append(pd.DataFrame(rows))
        elif group_files[0].endswith('.parquet'):
            for file in group_files:
                data_list.append(pd.read_parquet(folder + "/" + file))
        elif group_files[0].endswith('.csv'):
            for file in group_files:
                data_list.append(pd.read_csv(folder + "/" + file, sep=sep))
        else:
            print(f'FILES NOT EXIST')
            return
    data = pd.concat(data_list, axis=0, ignore_index=True)
    return data
```

```python
# Ваш исходный код до обработки
folder = Path('./results')
files = [f.name for f in folder.iterdir() if f.is_file()]
print(files)
pkl_files = [f for f in files if f.endswith('.pkl')]
print("Найденные .pkl файлы:", pkl_files)
pq_files = [f for f in files if f.endswith('.parquet')]
print("Найденные .parquet файлы:", pq_files)
csv_files = [f for f in files if f.endswith('.csv')]
print("Найденные .csv файлы:", csv_files)
```
